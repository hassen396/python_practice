```python
user_1 = {'x_position': 0, 'y_position': 0, 'speed': "medium"}
print(f'user one is {user_1}')
del user_1['y_position']
print(f'user one is {user_1}')

if user_1['speed']== 'slow':
    x_inc = 1
elif user_1['speed'] == 'medium':
    x_inc = 2
else:
    x_inc = 3

print(f"original x_position: {user_1['x_position']}")
user_1['x_position'] += x_inc
print(f"the new x_position is {user_1['x_position']}")
```
