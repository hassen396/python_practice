```python

class Solution:
    @staticmethod
    def length_of_longest_substring(s: str) -> int:
        if not s:
            return 0
        elif len(s) == 1:
            return 1
        p1 = 0
        p2 = 1
        maxx = 0
        subs = {s[p1]}
        while p2 < len(s):
            if s[p2] not in subs:
                subs.add(s[p2])
                p2 += 1
            else:
                maxx = max(maxx, len(subs))
                p1 += 1
                for sub1 in subs.copy():
                    if s[p2] == sub1:
                        subs.remove(sub1)
                        break
                p2 += 1
        return max(maxx, len(subs))


Solution.length_of_longest_substring("dvdf")
```

```python
def minimumIndex(capacity: list[int], itemSize: int) -> int:
    sm_index = {"index": -1, "value": None}
    for i in range(len(capacity)):
        if capacity[i] >= itemSize:
            if sm_index['value'] and capacity[i] < sm_index['value']:
                sm_index['index'] = i
                sm_index['value'] = capacity[i]
                continue
            elif sm_index['value'] and capacity[i] > sm_index['value']:
                continue
            sm_index['index'] = i
            sm_index['value'] = capacity[i]
    print(sm_index['index'])


minimumIndex([1, 5, 3, 7], 3)
```

```python
class Solution:
    @staticmethod
    def longestCommonPrefix(strs):
        if not strs: return ''
        com_pref = strs[0]
        for string in strs[1:]:
            if not com_pref: return ''
            com_pref = Solution.find_common_pref(com_pref, string)
        print('the common prefix is: ', com_pref)

    @staticmethod
    def find_common_pref(pref, string):
        sm_pref = ''
        for i in range(min(len(pref), len(string))):
            if pref[i] == string[i]:
                sm_pref += pref[i]
            else:
                break
        return sm_pref


Solution.longestCommonPrefix(["flower", "flow", "flight"])
```
