```python
import matplotlib.pyplot as plt
from pygame.surfarray import pixels_red

squares = [1, 4, 9, 16, 25]

fig, ax = plt.subplots()
ax.plot(squares, linewidth=3)

# Set chart title and label axes.
ax.set_title("Square Numbers", fontsize=24)
ax.set_xlabel("Value", fontsize=14)
ax.set_ylabel("Square of Value", fontsize=14)

# Set size of tick labels.
ax.tick_params(labelsize=14)
plt.show()
```

```python
from typing import List

def sortPeople(names: List[str], heights: List[int]) -> List[str]:
    n = len(names)
    name_and_height_tuples = [(heights[i], names[i]) for i in range(n)]
    print(sorted(name_and_height_tuples, reverse=True))
    return [element[1] for element in sorted(name_and_height_tuples, reverse=True)]
names = ["IEO","Sgizfdfrims","QTASHKQ","Vk","RPJOFYZUBFSIYp","EPCFFt","VOYGWWNCf","WSpmqvb"]
heights = [17233,32521,14087,42738,46669,65662,43204,8224]

res = sortPeople(names, heights)
print('res', res)
```

```python
def get_day_type(day):
    match day:
        case 'monday'| 'tuesday': return 'weekdayyy'
        case 'saterday' | 'sunday': return 'weekenddd'
        case _: return 'invalid day'

weekday = get_day_type('mondayy')
weekend = get_day_type('saterday')
print(weekday,'\n', weekend, sep='')
```
